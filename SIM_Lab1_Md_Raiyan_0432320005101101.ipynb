# Necessary Libraries

```python
import numpy as np
from sympy import Matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

# Scaler and Built In Functions

```python
# Task 1: Scalar & Array Operations with my own values
# Using scalar = 3.7 and a custom array of 6 elements

scalar = 3.7
arr = np.array([1.5, 2.8, 3.1, 4.6, 5.2, 6.9])  # my own array with 6 elements

# Trigonometric functions
sin_values = np.sin(arr)
cos_values = np.cos(arr)
tan_values = np.tan(arr)
# Normalize by max value before inverse trig functions to keep inputs inside [-1, 1]
asin_values = np.arcsin(np.clip(arr / np.max(arr), -1, 1))
acos_values = np.arccos(np.clip(arr / np.max(arr), -1, 1))
atan_values = np.arctan(arr)

# Exponential and logarithm
exp_values = np.exp(arr)
log_values = np.log(arr)  # Natural logarithm

# Absolute value and square root
abs_values = np.abs(arr)
sqrt_values = np.sqrt(arr)

# Remainder when divided by a scalar (2)
rem_values = np.remainder(arr, 2)

# Rounding functions
round_values = np.round(arr)
floor_values = np.floor(arr)
ceil_values = np.ceil(arr)

# Display results
print("scalar:", scalar)
print("array:", arr)
print("sin:", sin_values)
print("cos:", cos_values)
print("tan:", tan_values)
print("asin:", asin_values)
print("acos:", acos_values)
print("atan:", atan_values)
print("exp:", exp_values)
print("log:", log_values)
print("abs:", abs_values)
print("sqrt:", sqrt_values)
print("rem:", rem_values)
print("round:", round_values)
print("floor:", floor_values)
print("ceil:", ceil_values)
```

# 3×2 Matrix Calculations

```python
# Task 2: For 3x2 matrix
matrix = np.array([[1.0, 2.5],
                    [3.2, 4.8],
                    [5.1, 6.4]])

# Trigonometric functions
sin_values = np.sin(matrix)
cos_values = np.cos(matrix)
tan_values = np.tan(matrix)
asin_values = np.arcsin(np.clip(matrix / np.max(matrix), -1, 1))  # clipped to valid [-1, 1] domain
acos_values = np.arccos(np.clip(matrix / np.max(matrix), -1, 1))
atan_values = np.arctan(matrix)

# Exponential and logarithm
exp_values = np.exp(matrix)
log_values = np.log(matrix)  # Natural logarithm (log base e)

# Absolute value and square root
abs_values = np.abs(matrix)
sqrt_values = np.sqrt(matrix)

# Remainder when divided by a scalar (2)
rem_values = np.remainder(matrix, 2)

# Rounding functions
round_values = np.round(matrix)
floor_values = np.floor(matrix)
ceil_values = np.ceil(matrix)

# Display results
print("Matrix (3x2):\n", matrix)
print("\nsin:\n", sin_values)
print("\ncos:\n", cos_values)
print("\ntan:\n", tan_values)
print("\nasine:\n", asin_values)
print("\nacosine:\n", acos_values)
print("\natangent:\n", atan_values)
print("\nexp:\n", exp_values)
print("\nlog (natural):\n", log_values)
print("\nabs:\n", abs_values)
print("\nsqrt:\n", sqrt_values)
print("\nrem (remainder when divided by 2):\n", rem_values)
print("\nround:\n", round_values)
print("\nfloor:\n", floor_values)
print("\nceil:\n", ceil_values)
```

# Vector Functions
### Same Operations on a 2×3 Matrix

```python
# Task 3: Vector functions applied to a 2x3 matrix
mat = np.array([[2.1, 4.7, 1.3],
                 [9.4, 0.8, 6.2]])

# 1. Maximum value and its index (row, col)
max_value = np.max(mat)
max_index = np.unravel_index(np.argmax(mat), mat.shape)

# 2. Minimum value and its index (row, col)
min_value = np.min(mat)
min_index = np.unravel_index(np.argmin(mat), mat.shape)

# 3. "Length" of the matrix -> total number of elements
mat_length = mat.size

# 4. Sorting
sorted_rows_desc = np.sort(mat, axis=1)[:, ::-1]         # row-wise descending
sorted_all_desc = np.sort(mat.flatten())[::-1]           # entire matrix sorted, descending

# 5. Sum of elements
sum_values = np.sum(mat)

# 6. Product of elements
prod_values = np.prod(mat)

# 7. Median value
median_value = np.median(mat)

# 8. Mean value
mean_value = np.mean(mat)

# 9. Standard deviation
std_dev = np.std(mat)

# Display results
print("Matrix (2x3):\n", mat)
print("\nMax value:", max_value, "at index", max_index)
print("Min value:", min_value, "at index", min_index)
print("Total number of elements:", mat_length)
print("\nRow-wise descending sort:\n", sorted_rows_desc)
print("\nEntire matrix sorted descending:\n", sorted_all_desc)
print("\nSum of elements:", sum_values)
print("Product of elements:", prod_values)
print("Median value:", median_value)
print("Mean value:", mean_value)
print("Standard deviation:", std_dev)
```

# Bonus: Visualization & Random Number Generation

```python
# Bonus 1: Visualize the 2x3 matrix using a seaborn heatmap
plt.figure(figsize=(6, 4))
sns.heatmap(mat, annot=True, cmap="viridis", fmt=".2f", cbar=True)
plt.title("Heatmap of 2x3 Matrix Values")
plt.xlabel("Column")
plt.ylabel("Row")
plt.show()

# Bonus 2: Replace static inputs with randomly generated numbers
random_matrix = np.random.uniform(low=0, high=10, size=(2, 3))   # random floats between 0 and 10
random_ints = np.random.randint(low=1, high=20, size=6)          # random integers between 1 and 19

print("Randomly generated 2x3 matrix (uniform floats):\n", random_matrix)
print("\nRandomly generated integer array:\n", random_ints)

plt.figure(figsize=(6, 4))
sns.heatmap(random_matrix, annot=True, cmap="coolwarm", fmt=".2f", cbar=True)
plt.title("Heatmap of Randomly Generated 2x3 Matrix")
plt.xlabel("Column")
plt.ylabel("Row")
plt.show()
```
