# Hypothesis testing concepts: Two-Sample t-Test

## Import Necessary Libraries

```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
```

# Problem Solution

## Task 1: State H0 and H1 and assume equal variances (pooled).

```python
# Define the given data
data1 = np.array([18.4, 19.1, 17.9, 18.7, 18.3])  # Sample 1
data2 = np.array([20.2, 20.5, 20.1, 20.3, 20.4])  # Sample 2

# Pool t-test assumes equal variances
t_statistic, p_value = stats.ttest_ind(data1, data2, equal_var=True)
```

## Task 2: Compute sample means, sample variances (ddof = 1), and mean difference

```python
mean1 = np.mean(data1)
mean2 = np.mean(data2)

s1_sq = np.var(data1, ddof=1) # Sample variance for Data 1
s2_sq = np.var(data2, ddof=1) # Sample variance for Data 2

mean_diff = mean1 - mean2

print(f"Sample 1 Mean: {mean1:.2f}")
print(f"Sample 2 Mean: {mean2:.2f}")
print(f"Sample 1 Variance (ddof=1): {s1_sq:.4f}")
print(f"Sample 2 Variance (ddof=1): {s2_sq:.4f}")
print(f"Mean Difference (Data1 - Data2): {mean_diff:.2f}")
```

## Task 3: Calculate t-statistic, degrees of freedom, p-value

```python
# Sample sizes
n1, n2 = len(data1), len(data2)

# Degrees of freedom
df = n1 + n2 - 2

print(f"t-Statistic: {t_statistic:.2f}")
print(f"Degrees of Freedom: {df}")
print(f"p-Value: {p_value:.6f}")
```

## Task 4: 95% Confidence Interval for the mean difference

```python
# Pooled variance (assuming equal variances)
sp_sq = ((n1 - 1) * s1_sq + (n2 - 1) * s2_sq) / df

# Standard error of the difference between means
std_err = np.sqrt(sp_sq * (1/n1 + 1/n2))


# 95% Confidence Interval for the mean difference
ci = stats.t.interval(0.95, df, loc=mean_diff, scale=std_err)

print(f"95% Confidence Interval of Mean Difference: [{ci[0]:.2f}, {ci[1]:.2f}]")
```

## Task 5: Decision & Interpretation

```python
alpha = 0.05  # Significance level
if p_value < alpha:
    print("Reject the null hypothesis. The means of the two samples are significantly different.")
else:
    print("Fail to reject the null hypothesis. The means of the two samples are not significantly different.")
```
